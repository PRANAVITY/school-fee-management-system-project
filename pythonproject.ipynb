```python

class FeeAccount:
    def __init__(self, total_fee):
        self.total_fee = total_fee
        self.paid_fee = 0

    def pay_fee(self, amount):
        if amount <= 0:
            print("Please enter a valid amount.")
        elif amount > self.calculate_due():
            print("Amount is greater than the pending fee.")
        else:
            self.paid_fee += amount
            print("Fee paid successfully!")

    def calculate_due(self):
        return self.total_fee - self.paid_fee
lass Student:
    def __init__(self, student_id, name, total_fee):
        self.student_id = student_id
        self.name = name
        self.fee_account = FeeAccount(total_fee)

    def display_details(self):
        print("\n----- STUDENT FEE DETAILS -----")
        print("Student ID:", self.student_id)
        print("Student Name:", self.name)
        print("Total Fee: Rs.", self.fee_account.total_fee)
        print("Paid Fee: Rs.", self.fee_account.paid_fee)
        print("Pending Fee: Rs.", self.fee_account.calculate_due())


# Main Program

students = {}

while True:
    print("\n===== SCHOOL FEE MANAGEMENT SYSTEM =====")
    print("1. Register Student")
    print("2. Pay Fee")
    print("3. Calculate Pending Fee")
    print("4. Display Student Details")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        student_id = input("Enter Student ID: ")

        if student_id in students:
            print("Student already registered.")
            continue

        name = input("Enter Student Name: ")
        total_fee = float(input("Enter Total Fee: "))

        students[student_id] = Student(
            student_id, name, total_fee
        )

        print("Student registered successfully!")

    elif choice == "2":
        student_id = input("Enter Student ID: ")

        if student_id in students:
            amount = float(input("Enter Fee Amount: "))
            students[student_id].fee_account.pay_fee(amount)
        else:
            print("Student not found!")

    elif choice == "3":
        student_id = input("Enter Student ID: ")

        if student_id in students:
            due = students[student_id].fee_account.calculate_due()
            print("Pending Fee: Rs.", due)
        else:
            print("Student not found!")

    elif choice == "4":
        student_id = input("Enter Student ID: ")

        if student_id in students:
            students[student_id].display_details()
        else:
            print("Student not found!")

    elif choice == "5":
        print("Thank you for using School Fee Management System!")
        break

    else:
        print("Invalid choice! Please try again.")
```
